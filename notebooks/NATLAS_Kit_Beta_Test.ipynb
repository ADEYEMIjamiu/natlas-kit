## 2. Connect to N-ATLAS

```python
import os, getpass
os.environ["NATLAS_TESTER"]   = input("Your name: ")
os.environ["NATLAS_BASE_URL"] = input("Endpoint (ends with /v1): ").strip()
os.environ["NATLAS_API_KEY"]  = getpass.getpass("API key: ").strip()

from natlas_kit import NAtlas
nt = NAtlas()
print(nt.health())
```

## 3. Translate (try your own sentence too)

```python
for lang in ["yo", "ha", "ig"]:
    print(lang, "→", nt.translate("Please take this medicine twice a day after food.", target=lang))
```

## 4. Scam check: paste a real suspicious SMS or WhatsApp message you've received

```python
msg = input("Paste a message (or press Enter for an example): ") or "Your BVN has been blocked. Send your ATM PIN to 0803xxxxxxx to unlock it now."
import json; r = nt.classify_scam(msg); r.pop("raw", None)
print(json.dumps(r, indent=2, ensure_ascii=False))
```

## 5. Language detection & classification

```python
for t in ["How far, I don reach house.", "Ẹ kú àárọ̀, ṣé dáadáa ni?", "Ina kwana? Yaya aiki?", "Kedu ka ị mere?"]:
    print(nt.detect_language(t), "|", t)
print(nt.classify("Light never come for three days, transformer don spoil.", ["electricity", "water", "roads", "health"]))
```
